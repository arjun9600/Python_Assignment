# Banking Assignment — Functions, Lambda, map, filter, and reduce (Solved)

100 questions covering accounts, deposits, withdrawals, transfers, cards, loans,
KYC, fraud monitoring, reconciliation, and audit trails.

**Notes on approach**
- Each function states its input/output contract in a docstring and, where it
  helps, a type hint.
- `filter` is used for selection, `map` for transformation, and `reduce` only
  when there's no clearer built-in (e.g. `sum`) that already does the job.
- Functions that compute a value from their arguments avoid mutating any
  shared/mutable structure passed in -- they build and return new values.
- Every pipeline is tested against an empty collection to confirm it degrades
  safely instead of raising.

### Q31. Compose discount and tax functions for banking case 1; use sample reference ACC-1031 and explain the result.

```python
def apply_discount(amount, rate=0.1):
    return amount * (1 - rate)

def apply_tax(amount, rate=0.18):
    return amount * (1 + rate)

def compose(*funcs):
    def combined(x):
        for f in funcs:
            x = f(x)
        return x
    return combined


pipeline = compose(apply_discount, apply_tax)
print(f"[ACC-1031]", round(pipeline(1000), 2))


```

### Q35. Write a decorator that logs checkout calls for banking case 1; use sample reference ACC-1035 and explain the result.

```python
from functools import wraps

def logged(fn):
    @wraps(fn)
    def wrapper(*args, **kwargs):
        print(f"CALL {fn.__name__}({args}, {kwargs})")
        return fn(*args, **kwargs)
    return wrapper


@logged
def checkout(account_ref, amount):
    return f"[{account_ref}] charged {amount}"


print(checkout("ACC-1035", 500))
```

### Q39. Write a generator for low-balance accounts for banking case 1; use sample reference ACC-1039 and explain the result.

```python
def low_balance_accounts(accounts, threshold=1000):
    """Yield accounts below threshold, one at a time, lazily."""
    for a in accounts:
        if a["balance"] < threshold:
            yield a["ref"]


accounts = [{"ref": "ACC-1039", "balance": 500}, {"ref": "ACC-9999", "balance": 15000}]
for ref in low_balance_accounts(accounts):
    print("low balance:", ref)

```

### Q40. Implement recursive category total for banking case 1; use sample reference ACC-1040 and explain the result.

```python
def category_total(node):
    """Sum a nested category tree of fees, recursing into sub-categories."""
    total = node.get("amount", 0)
    for child in node.get("children", []):
        total += category_total(child)
    return total


tree = {"amount": 0, "children": [
    {"amount": 100},
    {"amount": 0, "children": [{"amount": 50}, {"amount": 25}]},
]}
print(f"[ACC-1040]", category_total(tree))


```

### Q41. Validate arguments with a wrapper for banking case 1; use sample reference ACC-1041 and explain the result.

```python
from functools import wraps

def positive_amount(fn):
    @wraps(fn)
    def wrapper(amount, *a, **kw):
        if amount <= 0:
            raise ValueError("amount must be positive")
        return fn(amount, *a, **kw)
    return wrapper


@positive_amount
def deposit(amount, account_ref="ACC-1041"):
    return f"[{account_ref}] deposited {amount}"


print(deposit(500))
try:
    deposit(-10)
except ValueError as e:
    print("REJECTED->", e)

```

```python
from functools import wraps

def retry(times=3):
    """A decorator FACTORY: retry(times=3) returns the actual decorator."""
    def decorator(fn):
        @wraps(fn)
        def wrapper(*a, **kw):
            last_exc = None
            for attempt in range(1, times + 1):
                try:
                    return fn(*a, **kw)
                except ConnectionError as exc:
                    last_exc = exc
                    print(f"  attempt {attempt} failed: {exc}")
            raise last_exc
        return wrapper
    return decorator


calls = {"n": 0}

@retry(times=3)
def call_gateway(account_ref="ACC-1073"):
    calls["n"] += 1
    if calls["n"] < 2:
        raise ConnectionError("gateway timeout")
    return f"[{account_ref}] payment confirmed"


print(call_gateway())

```
