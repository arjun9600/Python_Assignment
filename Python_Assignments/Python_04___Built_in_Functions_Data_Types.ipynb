### Q23. Removing false values using `filter()`
Filter out the `False` values from the list `[True, False, True, 0, 1, "", "Python"]`.

```python
data = [True, False, True, 0, 1, "", "Python"]
filtered_data = list(filter(None, data))
print(filtered_data)
```

### Q24. Using `map()`

### Q30. Multiple built-in functions
Create a program that takes a list of strings representing numbers, converts them to integers, and finds the sum and average.

```python
str_nums = ["10", "20", "30", "40"]
int_nums = list(map(int, str_nums))
total = sum(int_nums)
avg = total / len(int_nums)
print(f"List: {int_nums}, Sum: {total}, Average: {avg}")
```
