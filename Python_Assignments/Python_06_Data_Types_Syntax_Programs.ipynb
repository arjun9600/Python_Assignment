### Q1. Python program to find absolute value of a given number

```python
# Define a number
number = -15

# Find the absolute value using abs()
absolute_value = abs(number)

# Display the result
print(f"The absolute value of {number} is {absolute_value}")
```

### Q2. Python program to round a decimal number

```python
# Define a decimal number
decimal_number = 3.14159

# Round the number to the nearest integer
rounded_number = round(decimal_number)

# Display the result
print(f"The decimal number is {decimal_number}")
print(f"The rounded number to the nearest integer is {rounded_number}")
```

### Q3. Python program to round a number to 2 decimal places

```python
# Define a decimal number
decimal_number = 3.14159

# Round the number to 2 decimal places
rounded_two_places = round(decimal_number, 2)

# Display the result
print(f"The decimal number is {decimal_number}")
print(f"The rounded number to 2 decimal places is {rounded_two_places}")
```

### Q4. Python program to find the power of a number using pow()

```python
# Define base and exponent
base = 2
exponent = 3

# Calculate power using pow()
power_result = pow(base, exponent)

# Display the result
print(f"The power of {base} to the {exponent} is {power_result}")
```

### Q5. Python program to find the maximum value among three numbers using max()

```python
# Define three numbers
num1 = 10
num2 = 25
num3 = 15

# Find the maximum value using max()
maximum_value = max(num1, num2, num3)

# Display the result
print(f"The numbers are: {num1}, {num2}, {num3}")
print(f"The maximum value is {maximum_value}")
```

### Q6. Python program to find the minimum value among three numbers using min()

```python
# Define three numbers
num1 = 10
num2 = 25
num3 = 15

# Find the minimum value using min()
minimum_value = min(num1, num2, num3)

# Display the result
print(f"The numbers are: {num1}, {num2}, {num3}")
print(f"The minimum value is {minimum_value}")
```

### Q7. Python program to calculate the sum of elements in a list using sum()

```python
# Define a list of numbers
numbers_list = [1, 2, 3, 4, 5]

# Calculate the sum using sum()
list_sum = sum(numbers_list)

# Display the result
print(f"The list is: {numbers_list}")
print(f"The sum of elements in the list is {list_sum}")
```

### Q8. Python program to find length of a string using len()

```python
# Define a string
my_string = "Hello, World!"

# Find the length of the string using len()
string_length = len(my_string)

# Display the result
print(f"The string is: '{my_string}'")
print(f"The length of the string is {string_length}")
```

### Q10. Converting an integer into string using str()

```python
# Define an integer
my_integer = 12345

# Convert integer to string using str()
string_from_integer = str(my_integer)

# Display the result and its type
print(f"The integer is: {my_integer} (Type: {type(my_integer)})")
print(f"The string converted from integer is: '{string_from_integer}' (Type: {type(string_from_integer)})")
```

### Q11. Python program to convert a string containing a number into an integer using int()

```python
# Define a string containing a number
number_string = "67890"

# Convert string to integer using int()
integer_from_string = int(number_string)

# Display the result and its type
print(f"The string is: '{number_string}' (Type: {type(number_string)})")
print(f"The integer converted from string is: {integer_from_string} (Type: {type(integer_from_string)})")
```

### Q12. Converting integer into a floating point number using float()

```python
# Define an integer
an_integer = 789

# Convert integer to float using float()
float_from_integer = float(an_integer)

# Display the result and its type
print(f"The integer is: {an_integer} (Type: {type(an_integer)})")
print(f"The float converted from integer is: {float_from_integer} (Type: {type(float_from_integer)})")
```

### Q13. Python program to check the type of variable

```python
# Define variables of different types
var_int = 10
var_float = 3.14
var_string = "Python"
var_list = [1, 2, 3]
var_bool = True

# Check and display the type of each variable
print(f"Variable 'var_int': {var_int}, Type: {type(var_int)}")
print(f"Variable 'var_float': {var_float}, Type: {type(var_float)}")
print(f"Variable 'var_string': '{var_string}', Type: {type(var_string)}")
print(f"Variable 'var_list': {var_list}, Type: {type(var_list)}")
print(f"Variable 'var_bool': {var_bool}, Type: {type(var_bool)}")
```
