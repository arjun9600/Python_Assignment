## Python Math Module Questions - Solutions

This notebook provides solutions to 30 common questions related to Python's built-in `math` module.

```python
import math
```

### Q1. The Square root of a number

```python
num = 25
square_root = math.sqrt(num)
print(f"The square root of {num} is {square_root}")
```

### Q2. Cube root of a number

```python
num = 27
cube_root = num ** (1/3) # Using exponentiation operator
print(f"The cube root of {num} is {cube_root}")

# Alternatively, using math.pow()
cube_root_pow = math.pow(num, 1/3)
print(f"The cube root of {num} using math.pow is {cube_root_pow}")
```

### Q3. Factorial of a number

```python
num = 5
factorial = math.factorial(num)
print(f"The factorial of {num} is {factorial}")
```

### Q4. GCD of two numbers

```python
num1 = 48
num2 = 18
gcd = math.gcd(num1, num2)
print(f"The GCD of {num1} and {num2} is {gcd}")
```

### Q5. LCM of two numbers

```python
def lcm(a, b):
    return abs(a*b) // math.gcd(a, b)

num1 = 12
num2 = 18
lcm_val = lcm(num1, num2)
print(f"The LCM of {num1} and {num2} is {lcm_val}")
```

### Q6. Using `math.pow()`

```python
base = 2
exponent = 3
power = math.pow(base, exponent)
print(f"{base} raised to the power of {exponent} is {power}")
```

### Q7. Finding absolute value of negative decimal value

```python
num = -15.75
absolute_val = math.fabs(num)
print(f"The absolute value of {num} is {absolute_val}")
```

### Q8. Rounding a floating point number up (`math.ceil()`)

```python
num = 4.2
rounded_up = math.ceil(num)
print(f"Rounding {num} up gives {rounded_up}")
```

### Q9. Rounding a floating point number down (`math.floor()`)

```python
num = 4.9
rounded_down = math.floor(num)
print(f"Rounding {num} down gives {rounded_down}")
```

### Q10. Using `round()`

```python
num1 = 4.5
num2 = 4.2
num3 = 4.8
print(f"round({num1}) is {round(num1)}") # Rounds to nearest even number
print(f"round({num2}) is {round(num2)}")
print(f"round({num3}) is {round(num3)}")
```

### Q11. Using `math.fmod()`

```python
num = -10.5
divisor = 3
remainder = math.fmod(num, divisor)
print(f"The remainder of {num} divided by {divisor} using math.fmod() is {remainder}")
```

### Q12. Using `divmod()`

```python
num = 17
divisor = 5
quotient, remainder = divmod(num, divisor)
print(f"When {num} is divided by {divisor}, the quotient is {quotient} and the remainder is {remainder}")
```

### Q13. Calculating angles of degrees (using trigonometric functions)

```python
angle_deg = 30
angle_rad = math.radians(angle_deg)

sine_val = math.sin(angle_rad)
cosine_val = math.cos(angle_rad)
tangent_val = math.tan(angle_rad)

print(f"For an angle of {angle_deg} degrees ({angle_rad:.4f} radians):")
print(f"  Sine: {sine_val:.4f}")
print(f"  Cosine: {cosine_val:.4f}")
print(f"  Tangent: {tangent_val:.4f}")
```

### Q14. Converting angle degrees to radians

```python
angle_deg = 90
angle_rad = math.radians(angle_deg)
print(f"{angle_deg} degrees is equal to {angle_rad:.4f} radians")
```

### Q15. Converting from radians to degrees

```python
angle_rad = math.pi / 2 # Equivalent to 90 degrees
angle_deg = math.degrees(angle_rad)
print(f"{angle_rad:.4f} radians is equal to {angle_deg} degrees")
```

### Q16. Using logarithms (`math.log()`)

```python
num = 100
base = 10
log_val = math.log(num, base)
print(f"Log base {base} of {num} is {log_val}")
```

### Q17. Natural logarithm of a number (`math.log()` with base `e` or `math.log1p()` for `1+x`)

### Q18. Calculating e raised to the power of x (`math.exp()`)

```python
x = 2
exp_val = math.exp(x)
print(f"e raised to the power of {x} is {exp_val}")
```

### Q19. Value of pi and Euler's number

```python
print(f"Value of Pi (π): {math.pi}")
print(f"Value of Euler's number (e): {math.e}")
```

### Q20. Finding Distance between two points

```python
p1 = (1, 1)
p2 = (4, 5)
distance = math.dist(p1, p2) # Available in Python 3.8+
print(f"The distance between {p1} and {p2} is {distance}")

# Manual calculation for older Python versions:
distance_manual = math.sqrt((p2[0] - p1[0])**2 + (p2[1] - p1[1])**2)
print(f"The distance (manual) between {p1} and {p2} is {distance_manual}")
```

### Q21. Calculating the Hypotenuse of a right angle triangle (`math.hypot()`)

```python
side1 = 3
side2 = 4
hypotenuse = math.hypot(side1, side2)
print(f"The hypotenuse of a right triangle with sides {side1} and {side2} is {hypotenuse}")
```

### Q22. Combination (nCr) (`math.comb()`)

```python
n = 5
r = 2
combinations = math.comb(n, r)
print(f"The number of combinations of choosing {r} items from {n} is {combinations}")
```

### Q23. Permutation (nPr) (`math.perm()`)

```python
n = 5
r = 2
permutations = math.perm(n, r)
print(f"The number of permutations of choosing {r} items from {n} is {permutations}")
```

### Q24. Checking if a number is finite, infinite, or NaN

```python
val_finite = 100
val_infinite = float('inf')
val_nan = float('nan')

print(f"{val_finite} is finite: {math.isfinite(val_finite)}")
print(f"{val_finite} is infinite: {math.isinf(val_finite)}")
print(f"{val_finite} is NaN: {math.isnan(val_finite)}\n")

print(f"{val_infinite} is finite: {math.isfinite(val_infinite)}")
print(f"{val_infinite} is infinite: {math.isinf(val_infinite)}")
print(f"{val_infinite} is NaN: {math.isnan(val_infinite)}\n")

print(f"{val_nan} is finite: {math.isfinite(val_nan)}")
print(f"{val_nan} is infinite: {math.isinf(val_nan)}")
print(f"{val_nan} is NaN: {math.isnan(val_nan)}")
```

### Q25. Product of all numbers in a list (`math.prod()`)

```python
numbers = [1, 2, 3, 4, 5]
product = math.prod(numbers)
print(f"The product of numbers in the list {numbers} is {product}")
```

### Q26. Sum of square roots of numbers in a list

```python
numbers = [1, 4, 9, 16]
sum_sqrt = sum(math.sqrt(num) for num in numbers)
print(f"The sum of square roots of numbers in {numbers} is {sum_sqrt}")
```

### Q27. Finding the largest integer <= square root of given number

```python
num = 30
largest_int_le_sqrt = math.floor(math.sqrt(num))
print(f"The largest integer less than or equal to the square root of {num} is {largest_int_le_sqrt}")
```

### Q28. Compound interest using `math.pow()`

```python
principal = 1000
rate = 0.05 # 5%
time = 3 # years
n_compounded = 1 # annually

amount = principal * math.pow(1 + (rate / n_compounded), n_compounded * time)
compound_interest = amount - principal

print(f"Principal: ${principal}")
print(f"Interest Rate: {rate*100}%")
print(f"Time: {time} years")
print(f"Amount after compound interest: ${amount:.2f}")
print(f"Compound Interest: ${compound_interest:.2f}")
```

### Q29. Finding area, circumference of circle using `math.pi`

```python
radius = 7

area = math.pi * (radius ** 2)
circumference = 2 * math.pi * radius

print(f"For a circle with radius {radius}:")
print(f"  Area: {area:.2f}")
print(f"  Circumference: {circumference:.2f}")
```

### Q30. A Scientific Calculator (Demonstration of combining functions)

```python
def scientific_calculator(operation, *args):
    if operation == 'sqrt':
        return math.sqrt(args[0])
    elif operation == 'pow':
        return math.pow(args[0], args[1])
    elif operation == 'sin_deg':
        return math.sin(math.radians(args[0]))
    elif operation == 'log10':
        return math.log10(args[0])
    elif operation == 'hypot':
        return math.hypot(args[0], args[1])
    else:
        return "Invalid operation"

# Examples:
print(f"Square root of 64: {scientific_calculator('sqrt', 64)}")
print(f"2 to the power of 5: {scientific_calculator('pow', 2, 5)}")
print(f"Sine of 30 degrees: {scientific_calculator('sin_deg', 30):.4f}")
print(f"Log base 10 of 1000: {scientific_calculator('log10', 1000)}")
print(f"Hypotenuse of 5 and 12: {scientific_calculator('hypot', 5, 12)}")
```
