## Python 01 — String Indexing / Slicing

```python
# Q1. Create text = "Edukron Data Science and Data Engineering Batch 2026" and print len(text)
text = "Edukron Data Science and Data Engineering Batch 2026"
print(f"The string is: '{text}'")
print(f"The length of the string is: {len(text)}")
```
