## Class 2: *args and **kwargs Exercises


```python
# Assignment 1: Separate Positional and Keyword Arguments

def display_arguments(*args, **kwargs):
    positional_count = len(args)
    keyword_count = len(kwargs)

    print("Positional count:", positional_count)
    print("Keyword count:", keyword_count)
    print("Positional values:", args)
    print("Keyword values:", kwargs)


display_arguments(
    10,
    20,
    30,
    name="Rahul",
    city="Bangalore"
)
```

```python
# Assignment 2: Sum Only Numeric Positional Arguments

def calculate_numeric_total(*args):
    total = 0

    for value in args:
        if isinstance(value, (int, float)):
            total += value

    return total


result = calculate_numeric_total(10, "Python", 20.5, [1, 2], 30)
print("Numeric total:", result)
```

```python
# Assignment 3: Find Largest and Smallest Number

def find_min_max(*args):
    if len(args) == 0:
        return None, None

    largest = max(args)
    smallest = min(args)

    return largest, smallest


maximum, minimum = find_min_max(50, 10, 80, 25, 5)
print("Maximum:", maximum)
print("Minimum:", minimum)
```

```python
# Assignment 4: Student Marks Report

def student_report(name, *marks, **kwargs):
    total = sum(marks)
    average = total / len(marks) if marks else 0

    passed = all(mark >= 70 for mark in marks)

    print("Student:", name)
    print("Course:", kwargs.get("course", "Not provided"))
    print("City:", kwargs.get("city", "Not provided"))
    print("Total:", total)
    print("Average:", average)

    if passed:
        print("Result: PASS")
    else:
        print("Result: FAIL")


student_report(
    "Anil",
    80,
    75,
    90,
    65,
    course="Python",
    city="Hyderabad"
)
```

```python
# Assignment 5: Employee Salary Calculator

def calculate_salary(name, basic_salary, *args, **kwargs):
    allowance_total = sum(args)
    bonus = kwargs.get("bonus", 10000)
    tax_percentage = kwargs.get("tax", 10)

    gross_salary = basic_salary + allowance_total + bonus
    tax_amount = gross_salary * tax_percentage / 100
    net_salary = gross_salary - tax_amount

    print("Employee:", name)
    print("Gross salary:", gross_salary)
    print("Tax amount:", tax_amount)
    print("Net salary:", net_salary)


calculate_salary(
    "Rahul",
    50000,
    5000,
    3000,
    2000,
    bonus=10000,
    tax=10,
    department="IT"
)
```

```python
# Assignment 6: Count Data Types in *args

def count_data_types(*args):
    integer_count = 0
    string_count = 0
    list_count = 0
    other_count = 0

    for value in args:
        if type(value) is int:
            integer_count += 1
        elif isinstance(value, str):
            string_count += 1
        elif isinstance(value, list):
            list_count += 1
        else:
            other_count += 1

    print("Integers:", integer_count)
    print("Strings:", string_count)
    print("Lists:", list_count)
    print("Others:", other_count)


count_data_types(10, "Python", [1, 2], 20, "SQL", 3.5, True)
```

```python
# Assignment 7: Filter Even and Odd Numbers

def separate_even_odd(*numbers):
    even_numbers = []
    odd_numbers = []

    for number in numbers:
        if number % 2 == 0:
            even_numbers.append(number)
        else:
            odd_numbers.append(number)

    return even_numbers, odd_numbers


evens, odds = separate_even_odd(10, 15, 22, 31, 40, 47)
print("Even numbers:", evens)
print("Odd numbers:", odds)
```

```python
# Assignment 8: Search Employee Records

def search_employees(*employees, **kwargs):
    matched_employees = []

    for employee in employees:
        matches = all(
            employee.get(key) == value
            for key, value in kwargs.items()
        )

        if matches:
            matched_employees.append(employee)

    return matched_employees


employee1 = {
    "name": "Rahul",
    "department": "IT",
    "city": "Bangalore"
}

employee2 = {
    "name": "Anil",
    "department": "HR",
    "city": "Hyderabad"
}

employee3 = {
    "name": "Priya",
    "department": "IT",
    "city": "Bangalore"
}

results = search_employees(
    employee1,
    employee2,
    employee3,
    department="IT",
    city="Bangalore"
)

print(results)
```

```python
# Assignment 9: Shopping Cart Total

def checkout(*prices, **kwargs):
    subtotal = sum(prices)

    discount_percentage = kwargs.get("discount", 0)
    tax_percentage = kwargs.get("tax", 10)

    discount_amount = subtotal * discount_percentage / 100
    discounted_amount = subtotal - discount_amount

    tax_amount = discounted_amount * tax_percentage / 100
    final_amount = discounted_amount + tax_amount

    print("Customer:", kwargs.get("customer", "Guest"))
    print("Subtotal:", round(subtotal, 2))
    print("Discount:", round(discount_amount, 2))
    print("Tax:", round(tax_amount, 2))
    print("Final amount:", round(final_amount, 2))



checkout(
    1000,
    500,
    250,
    customer="Rahul",
    discount=10,
    tax=18
)
```

```python
# Assignment 11: Validate Required Keyword Arguments

def register_user(**kwargs):
    required_fields = ["name", "email", "password"]

    missing_fields = [
        field
        for field in required_fields
        if field not in kwargs
    ]

    if missing_fields:
        print("Missing fields:", missing_fields)
        return False

    print("Registration successful")
    return True


register_user(
    name="Rahul",
    email="rahul@example.com"
)
```

```python
# Assignment 12: Create a Dynamic SQL SELECT Query

def build_select_query(table, *columns, **filters):
    selected_columns = ", ".join(columns) if columns else "*"

    query = f"SELECT {selected_columns} FROM {table}"

    if filters:
        conditions = []

        for key, value in filters.items():
            conditions.append(f"{key}='{value}'")

        query += " WHERE " + " AND ".join(conditions)

    return query


query = build_select_query(
    "Employee",
    "name",
    "salary",
    department="IT",
    city="Bangalore"
)

print(query)
```

```python
# Assignment 13: Function Forwarding with *args and **kwargs

def execute_function(func, *args, **kwargs):
    print("Positional arguments:", args)
    print("Keyword arguments:", kwargs)

    result = func(*args, **kwargs)

    return result


def create_employee(name, age, department="IT", city="Bangalore"):
    return {
        "name": name,
        "age": age,
        "department": department,
        "city": city
    }


employee = execute_function(
    create_employee,
    "Rahul",
    28,
    department="Data Engineering",
    city="Hyderabad"
)

print(employee)
```

```python
# Assignment 14: Decorator with *args and **kwargs

def log_function_call(func):

    def wrapper(*args, **kwargs):
        print("Function name:", func.__name__)
        print("Positional:", args)
        print("Keyword:", kwargs)

        result = func(*args, **kwargs)

        print("Result:", result)

        return result
    
    return wrapper

@log_function_call
def calculate_bill(price, quantity, discount=0):
    total = price * quantity
    discount_amount = total * discount / 100
    return total - discount_amount


calculate_bill(500, 3, discount=10)
```

```python
# Assignment 15: Group Numbers Based on Conditions

def classify_numbers(*numbers, **kwargs):
    lower_limit = kwargs.get("lower_limit", 0)
    upper_limit = kwargs.get("upper_limit", 100)

    # ensure limits are in correct order
    if lower_limit > upper_limit:
        lower_limit, upper_limit = upper_limit, lower_limit

    below = []
    within = []
    above = []

    for number in numbers:
        if number < lower_limit:
            below.append(number)
        elif number > upper_limit:
            above.append(number)
        else:
            within.append(number)

    return {
        "below": below,
        "within": within,
        "above": above
    }


classification = classify_numbers(
    5,
    15,
    25,
    35,
    45,
    lower_limit=20,
    upper_limit=40
)

print(classification)
```

```python
# Assignment 16: Merge Multiple Dictionaries

def merge_dictionaries(*dictionaries, **overrides):
    merged = {}

    for dictionary in dictionaries:
        if isinstance(dictionary, dict):
            merged.update(dictionary)

    merged.update(overrides)

    return merged


result = merge_dictionaries(
    {"name": "Rahul", "city": "Bangalore"},
    {"department": "IT", "experience": 3},
    {"experience": 4},
    city="Hyderabad",
    salary=80000
)

print(result)
```

```python
# Assignment 17: ETL Pipeline Summary

import math


def etl_summary(source, destination, *tables, **kwargs):
    row_counts = kwargs.get("row_counts", {})
    batch_size = kwargs.get("batch_size", 1000)

    table_count = len(tables)

    total_rows = sum(
        row_counts.get(table, 0)
        for table in tables
    )

    batches = math.ceil(total_rows / batch_size) if total_rows else 0

    print("Source:", source)
    print("Destination:", destination)
    print("Tables:", tables)
    print("Table count:", table_count)
    print("Total rows:", total_rows)
    print("Required batches:", batches)


etl_summary(
    "SQL Server",
    "ADLS Gen2",
    "Customer",
    "Orders",
    "Products",
    row_counts={
        "Customer": 12000,
        "Orders": 55000,
        "Products": 3000
    },
    batch_size=10000
)
```

```python
# Assignment 18: Role-Based Permission Checker

def check_permissions(*actions, **kwargs):
    role_permissions = {
        "admin": {"read", "write", "update", "delete"},
        "editor": {"read", "write", "update"},
        "viewer": {"read"}
    }

    role = kwargs.get("role", "viewer")
    allowed_actions = role_permissions.get(role, set())

    print("User:", kwargs.get("username", "Unknown"))
    print("Role:", role)

    for action in actions:
        if action in allowed_actions:
            print(action, ": ALLOWED")
        else:
            print(action, ": DENIED")


check_permissions(
    "read",
    "update",
    "delete",
    username="Rahul",
    role="editor"
)
```

```python
# Assignment 19: Dynamic Object Creation

class Employee:

    def __init__(self, *skills, **details):
        self.skills = list(skills)

        for key, value in details.items():
            setattr(self, key, value)

    def display(self):
        print("Name:", getattr(self, "name", "Not provided"))
        print("Department:", getattr(self, "department", "Not provided"))
        print("Skills:", self.skills)


employee = Employee(
    "Python",
    "SQL",
    "Azure",
    name="Rahul",
    department="Data Engineering",
    experience=5
)

employee.display()
```

```python
# Assignment 20: Complex Order Processor

def process_order(order_id, *products, **kwargs):
    subtotal = 0
    total_quantity = 0

    for product in products:
        if not isinstance(product, dict):
            continue

        price = product.get("price", 0)
        quantity = product.get("quantity", 0)

        product_total = price * quantity

        subtotal += product_total
        total_quantity += quantity

    discount_percentage = kwargs.get("discount", 0)
    tax_percentage = kwargs.get("tax", 0)

    discount_amount = subtotal * discount_percentage / 100
    amount_after_discount = subtotal - discount_amount

    tax_amount = amount_after_discount * tax_percentage / 100
    final_amount = amount_after_discount + tax_amount

    print("Order ID:", order_id)
    print("Customer:", kwargs.get("customer", "Guest"))
    print("Payment mode:", kwargs.get("payment_mode", "Not provided"))
    print("Total quantity:", total_quantity)
    print("Subtotal:", round(subtotal, 2))
    print("Discount amount:", round(discount_amount, 2))
    print("Tax amount:", round(tax_amount, 2))
    print("Final amount:", round(final_amount, 2))


process_order(
    "ORD-1001",
    {
        "name": "Laptop",
        "price": 50000,
        "quantity": 1
    },
    {
        "name": "Mouse",
        "price": 1000,
        "quantity": 2
    },
    {
        "name": "Keyboard",
        "price": 2000,
        "quantity": 1
    },
    customer="Rahul",
    payment_mode="UPI",
    discount=10,
    tax=18
)
```

```python
'''1. Using filter() and map():

1. Extract all even numbers.
2. Calculate the square of every even number.
3. Return the squared values as a list.

'''

numbers = [12, 45, 67, 89, 23, 56, 91, 34, 78, 100]

even = filter(lambda x: x % 2 == 0, numbers)
result = list(map(lambda x: x ** 2, even))

print(result)
```

```python
'''2. Using filter() and reduce():

1. Extract all odd numbers.
2. Calculate their total sum.

'''

numbers = [12, 45, 67, 89, 23, 56, 91, 34, 78, 100]

odd = filter(lambda x: x % 2 != 0, numbers)
total = reduce(lambda a, b: a + b, odd)

print(total)
```

```python
'''3. Using filter() and map():

1. Extract numbers greater than 50.
2. Multiply every selected number by 2.

'''

numbers = [10, 25, 40, 55, 70, 85, 100]

greater = filter(lambda x: x > 50, numbers)
result = list(map(lambda x: x * 2, greater))

print(result)
```

```python
'''4. Using filter(), map(), and reduce():

1. Extract positive numbers.
2. Calculate the square of every positive number.
3. Calculate the total sum of the squared values.

'''

numbers = [-15, 20, -35, 40, 50, -10, 60]

positive = filter(lambda x: x > 0, numbers)
squares = list(map(lambda x: x ** 2, positive))
total = reduce(lambda a, b: a + b, squares)

print(squares)
print(total)
```

```python
'''5. Using filter() and map():

1. Extract words containing more than five characters.
2. Convert the selected words to uppercase.

'''

words = ["python", "azure", "ai", "developer", "cloud", "analytics", "spark"]

result = list(map(lambda x: x.upper(),filter(lambda x: len(x) > 5, words)))

print(result)
```

```python
'''6. Using filter() and map():

1. Extract names beginning with "A".
2. Convert the selected names to lowercase.

'''

names = ["Anil", "Rahul", "Ajay", "Meena", "Arjun", "Kiran", "Akhil"]

result = list(map(lambda x: x.lower(),filter(lambda x: x.startswith("A"), names)))

print(result)
```

```python
'''7. Using filter() and reduce():

1. Extract numbers divisible by both 3 and 5.
2. Calculate their total sum.

'''

numbers = [3, 5, 10, 15, 18, 30, 42, 45, 60, 75]

nums = filter(lambda x: x % 3 == 0 and x % 5 == 0, numbers)
total = reduce(lambda a, b: a + b, nums)

print(total)
```

```python
'''8. Using filter() and reduce():

1. Extract palindrome words.
2. Join them into one string separated by " | ".

'''

words = ["madam", "python", "level", "azure", "radar", "cloud", "civic"]

palindrome = filter(lambda x: x == x[::-1], words)
result = reduce(lambda a, b: a + " | " + b, palindrome)

print(result)
```

```python
'''9. Using filter(), map(), and reduce():

1. Extract temperatures greater than 30°C.
2. Convert the selected values into Fahrenheit.
3. Calculate the average Fahrenheit temperature.

'''

temperatures = [18, 24, 30, 35, 40, 12, 28, 33, 15, 42]

temp = list(map(lambda c: (c * 9/5) + 32,filter(lambda c: c > 30, temperatures)))

total = reduce(lambda a, b: a + b, temp)
average = total / len(temp)

print(temp)
print(average)
```

```python

'''20. Using filter(), map(), and reduce():

1. Extract active salespeople from the South region.
2. Calculate a 5% incentive for each selected salesperson.
3. Create a list containing salesperson name, sales amount, and incentive.
4. Calculate the total incentive payable.

'''

sales_records = [
    {"salesperson": "Rahul", "region": "South", "sales": 45000, "active": True},
    {"salesperson": "Anil", "region": "North", "sales": 65000, "active": True},
    {"salesperson": "Sneha", "region": "South", "sales": 72000, "active": False},
    {"salesperson": "Kiran", "region": "South", "sales": 82000, "active": True},
    {"salesperson": "Meena", "region": "West", "sales": 55000, "active": True},
    {"salesperson": "Arjun", "region": "South", "sales": 98000, "active": True}
]

selected = filter(
    lambda s: s["region"] == "South" and s["active"],
    sales_records
)

result = list(map(
    lambda s: {
        "salesperson": s["salesperson"],
        "sales": s["sales"],
        "incentive": s["sales"] * 0.05
    },
    selected
))

total = reduce(lambda a, b: a + b["incentive"], result, 0)

print(result)
print(total)
```
