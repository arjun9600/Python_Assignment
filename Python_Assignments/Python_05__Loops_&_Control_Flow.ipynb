### Q1. Adult age check with if/else

```python
# Q1. Adult age check with if/else
age = 20

if age >= 18:
    print("You are an adult.")
else:
    print("You are not yet an adult.")
```

### Q2. Even or odd check using modulus

```python
# Q2. Even or odd check using modulus
number = 7

if number % 2 == 0:
    print(f"{number} is an even number.")
else:
    print(f"{number} is an odd number.")
```

### Q3. Grade calculator using if/elif/else

```python
# Q3. Grade calculator using if/elif/else
score = 85

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
elif score >= 60:
    print("Grade: D")
else:
    print("Grade: F")
```

### Q4. Compare two numbers and print relation

```python
# Q4. Compare two numbers and print relation
num1 = 15
num2 = 10

if num1 > num2:
    print(f"{num1} is greater than {num2}.")
elif num1 < num2:
    print(f"{num1} is less than {num2}.")
else:
    print(f"{num1} is equal to {num2}.")
```

### Q11. Membership check using in with a list

```python
# Q11. Membership check using in with a list
fruits = ["apple", "banana", "cherry"]
item = "banana"

if item in fruits:
    print(f"{item} is in the list.")
else:
    print(f"{item} is not in the list.")
```

### Q12. Ternary / conditional expression

### Q13. for loop using range()

```python
# Q13. for loop using range()
print("Numbers from 0 to 4:")
for i in range(5):
    print(i)
```

### Q14. while loop practice

```python
# Q14. while loop practice
count = 0
print("Counting to 3:")
while count < 3:
    print(count)
    count += 1
```

### Q15. Loop control using break and/or continue

```python
# Q15. Loop control using break and/or continue
print("Demonstrating break:")
for i in range(10):
    if i == 5:
        print("Breaking loop at 5")
        break
    print(i)

print("\nDemonstrating continue:")
for i in range(5):
    if i == 2:
        print("Skipping 2")
        continue
    print(i)
```

### Q16. Nested loops practice

```python
# Q16. Nested loops practice
print("Multiplication table up to 3x3:")
for i in range(1, 4):
    for j in range(1, 4):
        print(f"{i} * {j} = {i*j}")
```
