## Q1. Create a nested bank/customer list structure (branches → customers → [ID, name, account type, balance])

```python
bank = [
    # Branch 1: Downtown Branch
    [
        # Customer 1
        ['C001', 'Alice Smith', 'Savings', 1500.00],
        # Customer 2
        ['C002', 'Bob Johnson', 'Checking', 3200.50],
        # Customer 3
        ['C003', 'Charlie Brown', 'Loan', -5000.00] # Negative balance for loan
    ],
    # Branch 2: Uptown Branch
    [
        # Customer 4
        ['C004', 'Diana Prince', 'Savings', 7800.25],
        # Customer 5
        ['C005', 'Clark Kent', 'Checking', 1200.75]
    ]
]

print("Bank structure created:")
# Optionally, display the entire bank structure
# print(bank)
```

## Q2. Access top-level bank branches using `bank[0]` / `bank[1]`

```python
print("\nAccessing Branch 1 (Downtown Branch):")
print(bank[0])

print("\nAccessing Branch 2 (Uptown Branch):")
print(bank[1])
```

## Q3. Access a complete customer record using multi-level indexing

```python
# Accessing Alice Smith's record (Branch 1, Customer 1)
alice_record = bank[0][0]
print("\nAlice Smith's complete record:")
print(alice_record)

# Accessing Clark Kent's record (Branch 2, Customer 2)
clark_record = bank[1][1]
print("\nClark Kent's complete record:")
print(clark_record)
```

## Q4. Extract a customer ID using nested indexing

```python
# Extracting Alice Smith's ID (Branch 1, Customer 1, ID is at index 0)
alice_id = bank[0][0][0]
print("\nAlice Smith's ID:")
print(alice_id)

# Extracting Diana Prince's ID (Branch 2, Customer 1, ID is at index 0)
diana_id = bank[1][0][0]
print("\nDiana Prince's ID:")
print(diana_id)
```

## Q5. Extract a customer name using nested indexing

```python
# Extracting Bob Johnson's name (Branch 1, Customer 2, Name is at index 1)
bob_name = bank[0][1][1]
print("\nBob Johnson's name:")
print(bob_name)

# Extracting Clark Kent's name (Branch 2, Customer 2, Name is at index 1)
clark_name = bank[1][1][1]
print("\nClark Kent's name:")
print(clark_name)
```

## Q6. Extract an account type using nested indexing

```python
# Extracting Charlie Brown's account type (Branch 1, Customer 3, Account type is at index 2)
charlie_account_type = bank[0][2][2]
print("\nCharlie Brown's account type:")
print(charlie_account_type)

# Extracting Diana Prince's account type (Branch 2, Customer 1, Account type is at index 2)
diana_account_type = bank[1][0][2]
print("\nDiana Prince's account type:")
print(diana_account_type)
```

## Q7. Extract a balance / loan amount using nested indexing

```python
# Extracting Alice Smith's balance (Branch 1, Customer 1, Balance is at index 3)
alice_balance = bank[0][0][3]
print("\nAlice Smith's balance:")
print(alice_balance)

# Extracting Charlie Brown's loan amount (Branch 1, Customer 3, Balance is at index 3)
charlie_loan = bank[0][2][3]
print("\nCharlie Brown's loan amount:")
print(charlie_loan)
```

```python
# Get Bob Johnson's balance
bob_balance = bank[0][1][3]
print(f"\nBob Johnson's balance: {bob_balance}")

# Get Diana Prince's name
diana_name = bank[1][0][1]
print(f"Diana Prince's name: {diana_name}")

# Get Clark Kent's ID
clark_id = bank[1][1][0]
print(f"Clark Kent's ID: {clark_id}")
```
