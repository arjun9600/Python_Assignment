# Banking Assignment — Regular Expressions (Solved)

100 questions covering accounts, deposits, withdrawals, transfers, cards, loans,
KYC, fraud monitoring, reconciliation, and audit trails.

**Notes on approach**
- All patterns use raw strings (`r"..."`) so backslashes stay literal.
- `fullmatch` is used for whole-field validation; `search`/`findall` for
  extraction from within a larger text.
- Named groups (`(?P<name>...)`) are used wherever a match needs to be pulled
  apart into meaningful pieces.
- Regex alone is never treated as full semantic validation (e.g. a date regex
  confirms *shape*, not that the calendar date actually exists) or as
  authorization -- those checks are called out explicitly where relevant.

### Q1. Find a account code for banking case 1; use sample reference ACC-1001 and explain the result.

## Medium — Reusable components and multi-step workflows (Q31-Q70)

### Q31. Parse a receipt line into named groups for banking case 1; use sample reference ACC-1031 and explain the result.

```python
import re

line = f"ACC-1031 | 3 | 249.50"
pattern = re.compile(r"^(?P<ref>[A-Z]{3}-\d{4})\s*\|\s*(?P<qty>\d+)\s*\|\s*(?P<amount>\d+(?:\.\d{2})?)$")
m = pattern.fullmatch(line)
print(m.groupdict() if m else "invalid")

```

### Q32. Validate an account reference with optional category prefix for banking case 1; use sample reference ACC-1032 and explain the result.

```python
import re

pattern = re.compile(r"(?:SAV-|CUR-|LOAN-)?ACC-\d{4}")
for value in ["ACC-1032", "SAV-ACC-1032", "XX-ACC-1032", "ACC-12"]:
    print(f"{value!r} valid? {bool(pattern.fullmatch(value))}")

```

### Q33. Extract currency and amount for banking case 1; use sample reference ACC-1033 and explain the result.

```python
import re

text = f"[ACC-1033] charged INR 1500.00 and refunded USD 12.50"
pattern = re.compile(r"(?P<currency>[A-Z]{3})\s(?P<amount>\d+(?:\.\d{2})?)")
for m in pattern.finditer(text):
    print(m.groupdict())

```

### Q34. Normalize account holder phone numbers for banking case 1; use sample reference ACC-1034 and explain the result.

```python
import re

raw_numbers = ["9876543210", "+91 98765 43210", "091-9876543210"]
pattern = re.compile(r"(?:\+?91[- ]?)?0?(?P<number>[6-9]\d{9})")
for raw in raw_numbers:
    m = pattern.search(raw.replace(" ", ""))
    print(f"[ACC-1034] {raw!r} -> +91{m.group('number') if m else None}")

```
