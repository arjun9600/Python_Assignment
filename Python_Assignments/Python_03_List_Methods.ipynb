### Q1. Create a list of 5 employee names and add a new employee using append().

```python
employee_names = ['Alice', 'Bob', 'Charlie', 'David', 'Eve']
employee_names.append('Frank')
print(f"Employee names after append: {employee_names}")
```

### Q2. Create a list of account numbers and add three new account numbers using extend().

```python
account_numbers = [1001, 1002, 1003]
new_accounts = [1004, 1005, 1006]
account_numbers.extend(new_accounts)
print(f"Account numbers after extend: {account_numbers}")
```
