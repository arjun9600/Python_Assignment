### Q78. Q20. Using reduce(), calculate the sum of unique numbers only. Do not directly use sum(set(numbers)).

```python
from functools import reduce

numbers = [1, 2, 2, 3, 4, 4, 5]
def add_unique(acc, x):
    current_sum, seen_numbers = acc
    if x not in seen_numbers:
        current_sum += x
        seen_numbers.add(x)
    return current_sum, seen_numbers

initial_state = (0, set())
final_sum, _ = reduce(add_unique, numbers, initial_state)
print(final_sum)
```

```python
from functools import reduce

users = [
    {'name': 'User1', 'email': 'user1@example.com', 'active': True},
    {'name': 'User2', 'email': 'user2@test.org', 'active': True},
    {'name': 'User3', 'email': 'user3@domain.com', 'active': False},
    {'name': 'User4', 'email': 'user4@example.com', 'active': True}
]

active_com_users_count = reduce(lambda count, _: count + 1, filter(lambda u: u['active'] and u['email'].endswith('.com'), users), 0)
print(active_com_users_count)
```

### Q97. Q20. Using filter(), map(), and reduce(): Calculate the total value of items that are in stock and have a price greater than 50.

```python
from functools import reduce

items = [
    {'name': 'ItemA', 'price': 60, 'in_stock': True},
    {'name': 'ItemB', 'price': 40, 'in_stock': True},
    {'name': 'ItemC', 'price': 75, 'in_stock': False},
    {'name': 'ItemD', 'price': 120, 'in_stock': True}
]

total_value_filtered_items = reduce(lambda acc, price: acc + price, map(lambda item: item['price'], filter(lambda item: item['in_stock'] and item['price'] > 50, items)), 0)
print(total_value_filtered_items)
```
