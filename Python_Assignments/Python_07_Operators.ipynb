## Q4. Reverse a three digit number

```python
num = 123

d1 = num % 10          # last digit
num = num // 10
d2 = num % 10          # middle digit
num = num // 10
d3 = num % 10          # first digit

reversed_num = d1 * 100 + d2 * 10 + d3
print(f"Original number: 123, Reversed number: {reversed_num}")
```

## Q6. Demonstrate all Assignment Operators (=, +=, -=, *=, /=, //=, %=, **=)

```python
x = 10
print(f"Initial x = {x}")

x = 5  # Assignment
print(f"x = 5  -> x = {x}")

x += 3 # x = x + 3
print(f"x += 3 -> x = {x}")

x -= 2 # x = x - 2
print(f"x -= 2 -> x = {x}")

x *= 4 # x = x * 4
print(f"x *= 4 -> x = {x}")

x /= 2 # x = x / 2
print(f"x /= 2 -> x = {x}")

x = 10 # Reset x for integer division
x //= 3 # x = x // 3
print(f"x //= 3 -> x = {x}")

x = 10 # Reset x for modulo
x %= 3 # x = x % 3
print(f"x %= 3 -> x = {x}")

x = 2  # Reset x for exponentiation
x **= 3 # x = x ** 3
print(f"x **= 3 -> x = {x}")
```

## Q7. Update a bank account balance using compound

```python
balance = 1000
deposit = 200
withdrawal = 50
interest_rate = 0.05

print(f"Initial balance: ${balance}")

balance += deposit  # Deposit money
print(f"Balance after deposit: ${balance}")

balance -= withdrawal # Withdraw money
print(f"Balance after withdrawal: ${balance}")

balance *= (1 + interest_rate) # Add interest
print(f"Balance after interest: ${balance:.2f}")
```

## Q18. Left Shift and right shift

```python
num = 8  # Binary: 1000

print(f"Original number: {num} ({bin(num)})")

# Left shift by 1 (equivalent to multiplying by 2)
left_shift = num << 1
print(f"Left shift by 1: {left_shift} ({bin(left_shift)})")

# Right shift by 1 (equivalent to integer division by 2)
right_shift = num >> 1
print(f"Right shift by 1: {right_shift} ({bin(right_shift)})")
```

## Q20. Swap two numbers using XOR

```python
a = 5   # Binary: 0101
b = 10  # Binary: 1010

print(f"Before swap: a = {a}, b = {b}")

a = a ^ b  # a becomes 0101 ^ 1010 = 1111 (15)
b = a ^ b  # b becomes 1111 ^ 1010 = 0101 (5)
a = a ^ b  # a becomes 1111 ^ 0101 = 1010 (10)

print(f"After swap: a = {a}, b = {b}")
```

## Q21. Count the number of Set Bits(1s)

```python
num = 23 # Binary: 00010111

count = 0
temp_num = num
while temp_num > 0:
    temp_num &= (temp_num - 1) # Brian Kernighan's algorithm
    count += 1

print(f"The number {num} ({bin(num)}) has {count} set bits.")

# Alternative using bin().count('1')
print(f"Using bin().count('1'): {bin(num).count('1')} set bits.")
```

## Q22. The difference between `is` and `==`

```python
# == compares the values of two objects.
# is compares the identity (memory address) of two objects.

a = [1, 2, 3]
b = [1, 2, 3]
c = a

print(f"a == b: {a == b} (values are equal)")
print(f"a is b: {a is b} (different objects in memory)")
print(f"a == c: {a == c} (values are equal)")
print(f"a is c: {a is c} (same object in memory)")

x = 10
y = 10
print(f"x == y: {x == y}")
print(f"x is y: {x is y}") # For small integers, Python often interns them, so 'is' can be True
```

## Q24. Behaviour of `is not` with integer and lists

## Q28. Simple calculator using Arithmetic and Comparison operators

```python
def simple_calculator(num1, num2, operator):
    if operator == '+':
        return num1 + num2
    elif operator == '-':
        return num1 - num2
    elif operator == '*':
        return num1 * num2
    elif operator == '/':
        if num2 != 0:
            return num1 / num2
        else:
            return "Error: Division by zero"
    elif operator == '%':
        if num2 != 0:
            return num1 % num2
        else:
            return "Error: Modulo by zero"
    else:
        return "Invalid operator"

print(f"10 + 5 = {simple_calculator(10, 5, '+')}")
print(f"10 - 5 = {simple_calculator(10, 5, '-')}")
print(f"10 * 5 = {simple_calculator(10, 5, '*')}")
print(f"10 / 5 = {simple_calculator(10, 5, '/')}")
print(f"10 % 3 = {simple_calculator(10, 3, '%')}")
print(f"10 / 0 = {simple_calculator(10, 0, '/')}")
```

## Q29. Grading System using Comparison & Logical Operators

```python
def get_grade(score):
    if score >= 90 and score <= 100:
        return 'A'
    elif score >= 80 and score < 90:
        return 'B'
    elif score >= 70 and score < 80:
        return 'C'
    elif score >= 60 and score < 70:
        return 'D'
    elif score >= 0 and score < 60:
        return 'F'
    else:
        return 'Invalid Score'

print(f"Score 95: Grade {get_grade(95)}")
print(f"Score 82: Grade {get_grade(82)}")
print(f"Score 70: Grade {get_grade(70)}")
print(f"Score 65: Grade {get_grade(65)}")
print(f"Score 45: Grade {get_grade(45)}")
print(f"Score 105: Grade {get_grade(105)}")
print(f"Score -5: Grade {get_grade(-5)}")
```

## Q30. Menu - Driven Calculator using Different Operators

```python
def menu_driven_calculator():
    while True:
        print("\n--- Menu Driven Calculator ---")
        print("1. Arithmetic operations")
        print("2. Compare two numbers")
        print("3. Bitwise operations")
        print("4. Membership check (number in a list)")
        print("5. Identity check (is vs ==)")
        print("6. Exit")

        choice = input("Choose an option (1-6): ")

        if choice == "1":
            x = float(input("Enter first number: "))
            y = float(input("Enter second number: "))
            print(f"Sum: {x + y}, Diff: {x - y}, Product: {x * y}")
            if y != 0:
                print(f"Division: {x / y}, Modulus: {x % y}")
            else:
                print("Division/modulus skipped (divide by zero)")

        elif choice == "2":
            x = float(input("Enter first number: "))
            y = float(input("Enter second number: "))
            if x > y:
                print(f"{x} is greater than {y}")
            elif x < y:
                print(f"{y} is greater than {x}")
            else:
                print("Both numbers are equal")

        elif choice == "3":
            x = int(input("Enter first integer: "))
            y = int(input("Enter second integer: "))
            print(f"AND: {x & y}, OR: {x | y}, XOR: {x ^ y}")
            print(f"Left shift by 1: {x << 1}, Right shift by 1: {x >> 1}")

        elif choice == "4":
            data = [10, 20, 30, 40, 50]
            val = int(input(f"Enter a number to search in {data}: "))
            if val in data:
                print(f"{val} is present in the list")
            else:
                print(f"{val} is not present in the list")

        elif choice == "5":
            a = [1, 2, 3]
            b = [1, 2, 3]
            c = a
            print(f"a == b: {a == b}")
            print(f"a is b: {a is b}")
            print(f"a is c: {a is c}")

        elif choice == "6":
            print("Exiting calculator. Bye!")
            break

        else:
            print("Invalid choice, please try again.")


if __name__ == "__main__":
    menu_driven_calculator()
```
