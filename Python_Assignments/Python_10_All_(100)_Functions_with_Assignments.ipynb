### Q3. Calculator Using Functions

```python
def calculator(operation, num1, num2):
    """Performs basic arithmetic operations based on the given operation."""
    if operation == 'add':
        return num1 + num2
    elif operation == 'subtract':
        return num1 - num2
    elif operation == 'multiply':
        return num1 * num2
    elif operation == 'divide':
        if num2 != 0:
            return num1 / num2
        else:
            return "Error: Cannot divide by zero"
    else:
        return "Error: Invalid operation"

# Demonstration
print(f"10 + 5 = {calculator('add', 10, 5)}")
print(f"10 - 5 = {calculator('subtract', 10, 5)}")
print(f"10 * 5 = {calculator('multiply', 10, 5)}")
print(f"10 / 5 = {calculator('divide', 10, 5)}")
print(f"10 / 0 = {calculator('divide', 10, 0)}")
```

### Q8. Area and Circumference of a Circle

```python
import math

def calculate_circle_properties(radius):
    """Calculates the area and circumference of a circle."""
    area = math.pi * (radius ** 2)
    circumference = 2 * math.pi * radius
    return area, circumference

# Demonstration
circle_area, circle_circumference = calculate_circle_properties(5)
print(f"For a circle with radius 5:")
print(f"Area: {circle_area:.2f}")
print(f"Circumference: {circle_circumference:.2f}")
```

### Q12. Check Leap Year

```python
def is_leap_year(year):
    """Checks if a given year is a leap year."""
    if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
        return True
    else:
        return False

# Demonstration
year1 = 2024
year2 = 1900
year3 = 2000
print(f"{year1} is a leap year: {is_leap_year(year1)}")
print(f"{year2} is a leap year: {is_leap_year(year2)}")
print(f"{year3} is a leap year: {is_leap_year(year3)}")
```

### Q13. Factorial Using a Function

```python
def calculate_factorial(n):
    """Calculates the factorial of a non-negative integer."""
    if n < 0:
        return "Factorial is not defined for negative numbers"
    elif n == 0:
        return 1
    else:
        fact = 1
        for i in range(1, n + 1):
            fact *= i
        return fact

# Demonstration
num1 = 5
num2 = 0
num3 = -3
print(f"The factorial of {num1} is: {calculate_factorial(num1)}")
print(f"The factorial of {num2} is: {calculate_factorial(num2)}")
print(f"The factorial of {num3} is: {calculate_factorial(num3)}")
```

### Q20. Count Positive, Negative and Zero Values

```python
def count_numbers_types(numbers):
    """Counts positive, negative, and zero values in a list."""
    positive_count = 0
    negative_count = 0
    zero_count = 0
    for num in numbers:
        if num > 0:
            positive_count += 1
        elif num < 0:
            negative_count += 1
        else:
            zero_count += 1
    return positive_count, negative_count, zero_count

# Demonstration
number_list = [1, -2, 0, 5, -10, 0, 7]
pos, neg, zero = count_numbers_types(number_list)
print(f"In {number_list}:\nPositive: {pos}, Negative: {neg}, Zeros: {zero}")
```

### Q34. Prime Number Function

```python
def is_prime(number):
    """Checks if a given number is a prime number."""
    if number <= 1:
        return False
    for i in range(2, int(number**0.5) + 1):
        if number % i == 0:
            return False
    return True

# Demonstration
print(f"7 is prime: {is_prime(7)}")
print(f"10 is prime: {is_prime(10)}")
print(f"2 is prime: {is_prime(2)}")
print(f"1 is prime: {is_prime(1)}")
```

### Q35. Fibonacci Function

```python
def generate_fibonacci(n):
    """Generates a list of the first n Fibonacci numbers."""
    fib_sequence = []
    a, b = 0, 1
    while len(fib_sequence) < n:
        fib_sequence.append(a)
        a, b = b, a + b
    return fib_sequence

# Demonstration
fib_list = generate_fibonacci(10)
print(f"First 10 Fibonacci numbers: {fib_list}")

fib_list_5 = generate_fibonacci(5)
print(f"First 5 Fibonacci numbers: {fib_list_5}")
```

### Q91. Closure-Based Counter

```python
def create_counter():
    count = 0
    def increment():
        nonlocal count
        count += 1
        return count
    return increment

counter = create_counter()
print(counter(), counter(), counter())
```

### Q92. Closure-Based Bank Balance

```python
def bank_account(initial_balance):
    balance = initial_balance
    def transaction(amount):
        nonlocal balance
        balance += amount
        return balance
    return transaction

account = bank_account(1000)
print(f"Deposit 500: {account(500)}")
print(f"Withdraw 200: {account(-200)}")
```

### Q100. Complete Functional Programming Project

```python
from functools import reduce

# Data
products = [
    {"name": "Laptop", "price": 1000, "category": "Electronics"},
    {"name": "Mouse", "price": 50, "category": "Electronics"},
    {"name": "Shirt", "price": 20, "category": "Apparel"},
    {"name": "Phone", "price": 500, "category": "Electronics"}
]

# Pipeline
# 1. Filter Electronics
# 2. Map: Apply 10% discount to price
# 3. Reduce: Get total cost

electronics = filter(lambda p: p['category'] == 'Electronics', products)
discounted = map(lambda p: {**p, 'price': p['price'] * 0.9}, electronics)
total_cost = reduce(lambda acc, p: acc + p['price'], discounted, 0)

print(f"Total discounted cost of electronics: ${total_cost:.2f}")
```
